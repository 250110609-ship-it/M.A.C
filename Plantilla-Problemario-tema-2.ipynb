```python
t = sympy.Symbol('t', real=True, positive=True)
s = sympy.Symbol('s')
a = sympy.Symbol('a', real=True, positive=True)
```

```python
# Definición de la función y su condición inicial
# dy/dt + 3*y = 13*sin(2*t), con y(0) = 6
y0 = 6
# Transformada de Laplace del lado derecho: L{13*sin(2*t)}
RHS_s = sympy.laplace_transform(13*sympy.sin(2*t), t, s, noconds=True)

# Despejando Y(s) de: s*Y(s) - y(0) + 3*Y(s) = RHS_s
# Y(s)*(s + 3) = RHS_s + y0
Y_s = sympy.simplify((RHS_s + y0) / (s + 3))
Y_s
```

```python
# Inversa de Laplace para obtener la solución y(t)
y_t = sympy.inverse_laplace_transform(Y_s, s, t)

print("En frecuencia Y(s):")
display(Y_s)
print("En tiempo y(t):")
display(y_t)
```

```python
# Importando módulos necesarios
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import sympy
from scipy import integrate

# Imprimir con notación matemática
sympy.init_printing(use_latex='mathjax')
# Ejemplo de transformada de Laplace
# Defino las incógnitas
t = sympy.symbols("t", positive=True)
y = sympy.Function("y")

# Defino la ecuación diferencial
edo = 1*y(t).diff(t) + 4*y(t) - sympy.exp(-4*t)
sympy.Eq(edo, 0)
```

```python
# Símbolos adicionales
s, Y = sympy.symbols("s, Y", real=True)
# Calculo la transformada de Laplace
L_edo = sympy.laplace_transform(edo, t, s, noconds=True)
sympy.Eq(L_edo, 0)
```

```python
def laplace_transform_derivatives(e):
    """
    Evalúa las transformadas de Laplace de derivadas de funciones sin evaluar.
    """
    if isinstance(e, sympy.LaplaceTransform):
        if isinstance(e.args[0], sympy.Derivative):
            d, t, s = e.args
            n = len(d.args) - 1
            return ((s**n) * sympy.LaplaceTransform(d.args[0], t, s) -
                    sum([s**(n-i) * sympy.diff(d.args[0], t, i-1).subs(t, 0)
                         for i in range(1, n+1)]))

    if isinstance(e, (sympy.Add, sympy.Mul)):
        t_type = type(e)
        return t_type(*[laplace_transform_derivatives(arg) for arg in e.args])

    return e
```

```python
# Aplicamos la nueva función para evaluar las transformadas de Laplace de derivadas
L_edo_2 = laplace_transform_derivatives(L_edo)
sympy.Eq(L_edo_2, 0)
```

```python
# Reemplazamos la transformada de Laplace de y(t) por la incógnita Y
L_edo_3 = L_edo_2.subs(sympy.laplace_transform(y(t), t, s), Y)
sympy.Eq(L_edo_3, 0)
```

```python
# Definimos la condición inicial
ics = {y(0): 2}
ics
```

```python
# Aplicamos la condición inicial
L_edo_4 = L_edo_3.subs(ics)
L_edo_4
```

```python
# 7. Reemplazamos la transformada por Y usando el objeto exacto
L_edo_3 = L_edo_2.subs(sympy.LaplaceTransform(y(t), t, s), Y)

# 9. Aplicamos las condiciones iniciales
L_edo_4 = L_edo_3.subs(ics)

# 10. Despejamos Y y calculamos la inversa de Laplace
Y_sol = sympy.solve(L_edo_4, Y)[0]
display(sympy.Eq(Y, Y_sol))

y_sol = sympy.inverse_laplace_transform(Y_sol, s, t)
y_sol
```

```python
# Importando módulos necesarios
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import sympy
from scipy import integrate

# Imprimir con notación matemática
sympy.init_printing(use_latex='mathjax')
# Ejemplo de transformada de Laplace
# Defino las incógnitas
t = sympy.symbols("t", positive=True)
y = sympy.Function("y")

# Defino la ecuación diferencial
edo = y(t).diff(t, t) + y(t) - sympy.sqrt(2) * sympy.sin(sympy.sqrt(2)*t)
sympy.Eq(edo, 0)
```

```python
# Símbolos adicionales
s, Y = sympy.symbols("s, Y", real=True)
# Calculo la transformada de Laplace
L_edo = sympy.laplace_transform(edo, t, s, noconds=True)
sympy.Eq(L_edo, 0)
```

```python
def laplace_transform_derivatives(e):
    """
    Evalúa las transformadas de Laplace de derivadas de funciones sin evaluar.
    """
    if isinstance(e, sympy.LaplaceTransform):
        if isinstance(e.args[0], sympy.Derivative):
            d, t, s = e.args
            n = len(d.args) - 1
            return ((s**n) * sympy.LaplaceTransform(d.args[0], t, s) -
                    sum([s**(n-i) * sympy.diff(d.args[0], t, i-1).subs(t, 0)
                         for i in range(1, n+1)]))

    if isinstance(e, (sympy.Add, sympy.Mul)):
        t_type = type(e)
        return t_type(*[laplace_transform_derivatives(arg) for arg in e.args])

    return e
```

```python
# Aplicamos la nueva función para evaluar las transformadas de Laplace de derivadas
L_edo_2 = laplace_transform_derivatives(L_edo)
sympy.Eq(L_edo_2, 0)
```

```python
# Reemplazamos la transformada de Laplace de y(t) por la incógnita Y
L_edo_3 = L_edo_2.subs(sympy.LaplaceTransform(y(t), t, s), Y)
sympy.Eq(L_edo_3, 0)
```

```python
# Definimos las condiciones iniciales
ics = {y(0): 10, y(t).diff(t).subs(t, 0): -1}
ics
```

```python
# Aplicamos las condiciones iniciales
L_edo_4 = L_edo_3.subs(ics)
L_edo_4
```

```python
# Despejamos Y de la ecuación L_edo_4 == 0
Y_sol = sympy.solve(L_edo_4, Y)[0]
display(sympy.Eq(Y, Y_sol))

# Calculamos la inversa de la Transformada de Laplace para obtener y(t)
y_sol = sympy.inverse_laplace_transform(Y_sol, s, t)
y_sol
```
